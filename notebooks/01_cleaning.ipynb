```python
import pandas as pd

df = pd.read_csv('../data/cleaned/nba_cleaned.csv')

# Simplify combo positions to primary position
df['Pos'] = df['Pos'].str.split('-').str[0]

# Verify
print(df['Pos'].unique())
print(df['Pos'].value_counts())

# Save back
df.to_csv('../data/cleaned/nba_cleaned.csv', index=False)
print("Done!")
```
